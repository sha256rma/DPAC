```python
import pandas as pd

# Read each CSV file
bmx_df = pd.read_csv("BMX_D.csv")
bpq_df = pd.read_csv("BPQ_D.csv")
bpx_df = pd.read_csv("BPX_D.csv")
demo_df = pd.read_csv("DEMO_D.csv")
diq_df = pd.read_csv("DIQ_D.csv")
glu_df = pd.read_csv("GLU_D.csv")
mcq_df = pd.read_csv("MCQ_D.csv")
tchol_df = pd.read_csv("TCHOL_D.csv")
```

```python
# Data cleaning
def clean_data(df):
    # Drop rows with any missing values
    df.dropna(inplace=True)
    # Convert SEQN to int if it's not already
    if 'SEQN' in df.columns:
        df['SEQN'] = df['SEQN'].astype(int)
```

```python
# Clean each dataframe
clean_data(bmx_df)
clean_data(bpq_df)
clean_data(bpx_df)
clean_data(demo_df)
clean_data(diq_df)
clean_data(glu_df)
clean_data(mcq_df)
clean_data(tchol_df)
```

```python
# Merge the dataframes based on SEQN
merged_data = pd.merge(bmx_df, bpq_df, on='SEQN', how='inner')
merged_data = pd.merge(merged_data, bpx_df, on='SEQN', how='inner')
merged_data = pd.merge(merged_data, demo_df, on='SEQN', how='inner')
merged_data = pd.merge(merged_data, diq_df, on='SEQN', how='inner')
merged_data = pd.merge(merged_data, glu_df, on='SEQN', how='inner')
merged_data = pd.merge(merged_data, mcq_df, on='SEQN', how='inner')
merged_data = pd.merge(merged_data, tchol_df, on='SEQN', how='inner')
```

```python
# Calculate average of 'BPXSY1', 'BPXSY2', and 'BPXSY3' columns
merged_data['BPXSY_avg'] = merged_data[['BPXSY1', 'BPXSY2', 'BPXSY3']].mean(axis=1)

# Remove 'BPXSY1', 'BPXSY2', and 'BPXSY3' columns
merged_data.drop(['BPXSY1', 'BPXSY2', 'BPXSY3'], axis=1, inplace=True)
```

```python
# Columns to convert to int64
columns_to_convert = ['BPQ020', 'BPQ080', 'BPQ090D', 'RIAGENDR', 'RIDAGEYR', 'DIQ010', 'MCQ300C']

# Convert columns to int64
merged_data[columns_to_convert] = merged_data[columns_to_convert].astype('int64')
```

```python
# BPQ020,BPQ080,BPQ090D,DIQ010,MCQ300C: 1=yes; 2=no --change to--> 1=yes; 0=no

# Columns to replace values in
columns_to_replace = ['BPQ020', 'BPQ080', 'BPQ090D', 'DIQ010', 'MCQ300C']

# Replace all occurrences of 2 with 0 in specified columns
merged_data[columns_to_replace] = merged_data[columns_to_replace].replace(2, 0)
```

```python
# RIAGENDR: 1=Male; 2=Female --change to--> 0=Male; 1=Female

# Replace values in the 'RIAGENDR' column
merged_data['RIAGENDR'] = merged_data['RIAGENDR'].replace({1: 0, 2: 1})
```

```python
# Save the updated merged dataframe to a new CSV file
merged_data.to_csv("merged_data.csv", index=False)
```
